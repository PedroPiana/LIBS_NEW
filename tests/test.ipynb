```python
from collections import Counter
from pathlib import Path


def _resolve_dat_path(dat_path: str | Path) -> Path:
    path = Path(dat_path)
    if path.is_absolute() and path.exists():
        return path

    for base in (Path.cwd(), *Path.cwd().parents):
        candidate = base / path
        if candidate.exists():
            return candidate

    raise FileNotFoundError(f"Arquivo não encontrado: {dat_path}")


def listar_nomes_primeira_linha(dat_path: str | Path = "dataset/laser266.dat", ignorar_primeiro: bool = True) -> Counter[str]:
    """Lê a primeira linha do .dat e conta quantas vezes cada nome aparece."""
    arquivo = _resolve_dat_path(dat_path)

    with arquivo.open("r", encoding="utf-8", errors="ignore") as f:
        primeira_linha = f.readline().strip()

    if not primeira_linha:
        raise ValueError(f"O arquivo está vazio: {arquivo}")

    nomes = [item.strip() for item in primeira_linha.split(";") if item.strip()]
    if ignorar_primeiro and nomes:
        nomes = nomes[1:]

    contagem = Counter(nomes)

    print(f"Arquivo: {arquivo}")
    print("Nome -> quantidade")
    for nome, quantidade in contagem.most_common():
        print(f"{nome} -> {quantidade}")

    return contagem


contagem_nomes = listar_nomes_primeira_linha()
```
